```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
df = pd.read_csv("E:\\Documentos 2\\ML y DL\\insurance.csv")
```

```python
df.head()
```

```python
#variables de texto a variables numéricas
df["sex"] = df["sex"].map({
    "male": 0,
    "female": 1
})

df["smoker"] = df["smoker"].map({
    "no": 0,
    "yes": 1
})
df = pd.get_dummies(df, columns=["region"], dtype=int)#convertimos region utilizando variables dummy
```

```python
#variables que utilizaremos para realizar la predicción
X = df.drop("charges", axis=1)
y = df["charges"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
modelo = RandomForestRegressor(
    n_estimators=100,
    random_state=42
)
```

```python
modelo.fit(X_train, y_train)
```

```python
#estimar
y_pred = modelo.predict(X_test)
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(y_test, y_pred)
plt.xlabel("Gastos medicos reales")
plt.ylabel("Gastos medicos estimados")
plt.title("Random Forest - reales vs predicciones")
plt.show()
```

```python
importancias = pd.Series(
    modelo.feature_importances_,
    index=X.columns
)
print(importancias.sort_values(ascending=False))
```

```python
importancias.sort_values().plot(
    kind="barh",
    figsize=(7, 5)
)

plt.xlabel("Importancia")
plt.ylabel("Variable")
plt.title("Importancia de las variables")

plt.tight_layout()
plt.show()
```
